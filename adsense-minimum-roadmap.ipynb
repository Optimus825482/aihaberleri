# AIHaberleri.org — AdSense Minimum Uygunluk Yol Haritası

> Not: Bu çalışma, kullanıcı ifadesindeki "At sensin" ifadesini **Google AdSense** olarak yorumlayarak hazırlanmıştır.

## Amaç
Siteyi, minimum eforla AdSense başvurusuna uygun hale getirmek ve ilk onay olasılığını artırmak.

## 1) Mevcut Durum Özeti (Kısa)

### Güçlü Yanlar
- Geniş içerik altyapısı ve düzenli yayın akışı mevcut.
- Hakkımızda / İletişim / Gizlilik / Şartlar / Çerez gibi güven sayfaları mevcut.
- SEO temel taşları (robots, sitemap, metadata, structured data) uygulanmış.
- Çok dilli yapı (TR/EN) ve haber detay sayfaları olgun.

### Kritik Eksikler (AdSense için)
1. `public/ads.txt` yok.
2. AdSense script/slot entegrasyonu yok (reklam katmanı şu an devre dışı).
3. Çerez onayı var ama reklam kişiselleştirme/consent akışı AdSense odaklı değil.
4. Kurumsal iletişim bilgilerinde domain/e-posta tutarlılığı gözden geçirilmeli.
5. İçerik kalite eşiği (inceleme için yeterli özgün ve dengeli yayın hacmi) ölçülebilir KPI ile takip edilmiyor.

## 2) Minimum Gereksinim Listesi (Must-Have)

### A. Teknik
- [ ] `ads.txt` yayına alınmalı.
- [ ] AdSense hesap doğrulaması (site ownership + domain doğrulama) tamamlanmalı.
- [ ] Global layout’a AdSense script entegrasyonu kontrollü eklenmeli.
- [ ] En az 2-3 güvenli reklam yerleşimi (header altı, içerik içi, içerik sonu) tanımlanmalı.
- [ ] CLS/LCP etkisi düşük olacak şekilde lazy yükleme ve placeholder uygulanmalı.

### B. Politika/Uyum
- [ ] Gizlilik ve Çerez politikasında Google AdSense ve veri işleme metni açık belirtilmeli.
- [ ] Kullanıcıya reklam/analitik çerez tercihi verecek şekilde consent akışı güncellenmeli.
- [ ] Yasak içerik kategorileri için editoryal kontrol listesi netleştirilmeli.

### C. İçerik Kalitesi
- [ ] Son 30 günde düzenli yayın ritmi korunmalı.
- [ ] İnceleme öncesi yeterli sayıda özgün, derinlikli makale canlıda olmalı.
- [ ] Düşük kaliteli/çok kısa içerikler gözden geçirilmeli (thin content temizliği).

### D. Güven & Şeffaflık
- [ ] İletişim sayfasında marka-domain-e-posta tutarlılığı sağlanmalı.
- [ ] Hakkımızda sayfasında ekip/kurucu ve editoryal süreç net olmalı.
- [ ] Footer’da tüm yasal bağlantılar TR ve EN için erişilebilir olmalı.

## 3) 14 Günlük Minimum Yol Haritası

### Faz 1 — Uyum Tabanı (Gün 1-3)
1. `ads.txt` oluştur ve yayına al.
2. Gizlilik/Çerez metinlerine AdSense + çerez kullanım amacı ekle.
3. Consent bileşenini kategorik tercihe (zorunlu/analitik/reklam) genişlet.

**Çıktı:** Politika ve teknik zorunlular tamam.

### Faz 2 — Reklam Altyapısı (Gün 4-7)
1. AdSense script’i feature flag ile ekle (`NEXT_PUBLIC_ADSENSE_ENABLED`).
2. 2-3 reklam bileşeni oluştur (`AdSlot`), admin sayfalarında kapalı tut.
3. Performans ölç: LCP/CLS etkisini raporla, gerekirse slotları optimize et.

**Çıktı:** Reklam katmanı policy-safe ve performans kontrollü.

### Faz 3 — İçerik ve Güven Sinyalleri (Gün 8-11)
1. Düşük kaliteli içerik temizliği + önemli içeriklerin güncellenmesi.
2. TR/EN legal ve iletişim içerik tutarlılık kontrolü.
3. Site içi gezinme ve kategori keşfi güçlendirme (zaten güçlü, final polish).

**Çıktı:** Başvuru öncesi kalite eşiği.

### Faz 4 — Başvuru Hazırlığı (Gün 12-14)
1. Son teknik check: robots/sitemap/indexlenebilirlik.
2. Manuel QA: reklam yerleşimi, kırık link, mobil görünüm, 404/500 kontrolü.
3. AdSense başvurusu + 7-14 gün izleme planı.

**Çıktı:** Başvuruya hazır üretim ortamı.

## 4) Uygulama İş Listesi (Developer-Ready)

### İş Paketi WP-01 — ads.txt
- Dosya: `public/ads.txt`
- Kabul Kriteri: `https://domain.com/ads.txt` 200 dönmeli ve yayıncı satırı görünmeli.

### İş Paketi WP-02 — AdSense Script
- Dosya: `src/app/layout.tsx`
- Kabul Kriteri: Production’da script yüklenmeli, local/dev’de flag ile kapatılabilmeli.

### İş Paketi WP-03 — Ad Slot Component
- Dosya: `src/components/AdSlot.tsx` (yeni)
- Kabul Kriteri: `slot`, `format`, `responsive`, `className` parametreleriyle çalışmalı.

### İş Paketi WP-04 — Consent Güncellemesi
- Dosya: `src/components/CookieConsent.tsx`
- Kabul Kriteri: reklam tercihi reddedilirse kişiselleştirilmiş reklam scripti yüklenmemeli.

### İş Paketi WP-05 — Policy Metinleri
- Dosyalar: `src/app/privacy/page.tsx`, `src/app/cookies/page.tsx`
- Kabul Kriteri: AdSense veri işleme, üçüncü taraf reklam, tercih yönetimi açık şekilde yazılı olmalı.

## 5) Riskler ve Önlemler

1. **Thin Content Riski**
   - Önlem: Başvuru öncesi düşük kalite içerik temizliği + içerik güncelleme sprinti.
2. **Policy Mismatch Riski**
   - Önlem: Gizlilik/çerez metinlerini gerçek davranışla birebir hizala.
3. **Performans Düşüşü (Ads)**
   - Önlem: Lazy load + sadece görünür yerde reklam + CLS koruması.
4. **Güven Sinyali Tutarsızlığı**
   - Önlem: marka, domain, mail adresleri ve iletişim bilgilerinin tekilleştirilmesi.

## 6) Başarı Ölçütleri (Go/No-Go)

Başvuruya geçmeden önce hepsi TRUE olmalı:

- [ ] `ads.txt` canlı ve doğru.
- [ ] Legal sayfalar güncel ve erişilebilir.
- [ ] Reklam script/slot testleri tamam.
- [ ] Mobilde ve masaüstünde kritik UX sorunu yok.
- [ ] Son 30 gün içerik yayın ritmi korunuyor.
- [ ] Search Console’da kritik indeksleme hatası yok.

## 7) Hızlı Kazanım Sırası (Önerilen)

1. `ads.txt`
2. Consent + policy hizalama
3. AdSlot bileşeni ve kontrollü rollout
4. İçerik kalite sprinti
5. Başvuru

Bu sıra, en düşük eforla en yüksek başvuru etkisini hedefler.
